```python
# Filtrar los plans

# Ruta del archivo original en tu Mac
input_file_path = "/Users/gustavotitochoque/Documents/arequipa/current_Data/access.log"

# Ruta del nuevo archivo filtrado
output_file_path = "/Users/gustavotitochoque/Documents/arequipa/output/arequipa_filtrado.log"

# Palabras clave que queremos buscar en cada línea
filtros = ["/otp/routers/default/plan", "Arequipa Bus"]

# Leer el archivo línea por línea, filtrar y guardar
with open(input_file_path, "r", encoding="utf-8") as infile, open(output_file_path, "w", encoding="utf-8") as outfile:
    for line in infile:
        if all(filtro in line for filtro in filtros):  # Verifica si alguna palabra clave está en la línea
            outfile.write(line)  # Guarda la línea en el nuevo archivo

print(f"✅ Proceso completado. Archivo filtrado guardado en: {output_file_path}")
```

```python
import re
import csv
from urllib.parse import unquote

# Ruta de tu archivo de log
log_file_path = '/Users/gustavotitochoque/Documents/arequipa/output/arequipa_filtrado.log'
# Ruta donde quieres guardar el CSV
csv_output_path = '/Users/gustavotitochoque/Documents/arequipa/output/data.csv'

# Expresión regular para extraer la información.
# Se asume que tu línea respeta más o menos el formato:
# IP - - [fecha_hora] "GET /otp/routers/default/plan?fromPlace=...&toPlace=... HTTP/1.1" 200 ... "Trufi/.../userID/Arequipa Bus" "-"
pattern = re.compile(
    r'^(?P<ip>\S+)\s+-\s+-\s+\[(?P<datetime>[^\]]+)\]\s+'
    r'"GET\s+/otp/routers/default/plan\?fromPlace=(?P<fromPlace>[^&]+)&toPlace=(?P<toPlace>[^&]+).*"\s+'
    r'\d+\s+\d+\s+"-"\s+"Trufi/[^/]+/(?P<user>[^/]+)/Arequipa\s+Bus"\s+"-"'
)

# Abrimos el archivo de log en modo lectura y el CSV en modo escritura
with open(log_file_path, 'r', encoding='utf-8') as log_file, \
     open(csv_output_path, 'w', newline='', encoding='utf-8') as csv_file:

    writer = csv.writer(csv_file)
    # Escribimos la cabecera del CSV
    writer.writerow(['id', 'fecha_hora', 'originLat', 'originLng', 'destinyLat', 'destinyLng', 'userId'])
    
    id_counter = 1
    
    for line in log_file:
        match = pattern.search(line)
        if match:
            # Extraemos campos
            fecha_hora = match.group('datetime')        # Ej: 04/Feb/2025:14:05:13 +0000
            from_place = match.group('fromPlace')       # Ej: -16.3889322%2C-71.5492756
            to_place   = match.group('toPlace')         # Ej: -16.4541435%2C-71.4947172
            user_id    = match.group('user')            # Ej: 211e5678-f69f-482e-9810-1c5308f8f5e3
            
            # Decodificar por si hay %2C etc.
            from_decoded = unquote(from_place)  # "-16.3889322,-71.5492756"
            to_decoded   = unquote(to_place)    # "-16.4541435,-71.4947172"
            
            # Separar lat y lng
            try:
                originLat, originLng = from_decoded.split(',')
                destinyLat, destinyLng = to_decoded.split(',')
            except ValueError:
                # Por si acaso la línea no tiene el formato esperado
                continue
            
            # Escribimos la fila en el CSV
            writer.writerow([
                id_counter,
                fecha_hora,
                originLat,
                originLng,
                destinyLat,
                destinyLng,
                user_id
            ])
            
            id_counter += 1

print("Proceso completado. CSV creado en:", csv_output_path)
```
